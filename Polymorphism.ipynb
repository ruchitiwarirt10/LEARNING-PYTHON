```python
#Polymorphism-Polymorphism is a many forms-Poly-many,Morphism-forms
#Types of Polymorphism--1.Overriding-change-
                        #a.Method overriding-In Python,if same method present in parent
                                            #and child class and we create object of child 
                                            #class then that method in child class override
                                                                #parent class's method,
#2.Overloading-Above Capacity-a.Method Overloading,b.Operator Overloading

class parent:
    def display(self):
        print("its Parent Class")

class Child(parent):
    def display(self):
        print("its Child Class")

c=Child()
c.display()  #child class's display method override parent's display method.
```

```python
#Overloading-
##1..Method overloading-Not supported in Python

def add(a,b):
    print(a+b)
add(1,2)

def add(a,b,c):
    print(a+b+c)
add(1,2,3)
```

```python
#Method Overloading-Python not Support directly-(*)-multiple/n-number of (args)-arguments
def add(*args):
    print(sum(args))   #args=tuple()=(1,2,3,4,5,6,7,8,9)
                       #sum((1,2,3,4,5,6,7,8,9))
add(1,2)
add(1,2,3)
add(1,2,3,4,5,6,7,8,9)
```

```python
#2.Operator Overloading-
class Employee:
    def __init__(self,name,salary):     #Dunder Method or special method for operator overloading
        self.name=name
        self.__salary=salary

    def display(self):
        print("Employee name is ",self.name,"And salary is",self.__salary)

    def __add__(self,other):            #Dunder Method  #Double underscore method
        return self.__salary+other.__salary

e1=Employee("Yash",50000)
e1.display()

e2=Employee("Saurav",80000)
e2.display()

print(e1+e2)     #e1.__add__(e2)      #e1.salary+e2.salary
```
