```python
#type:-type function is a predefine function, that returns the data type or class of particular object.
```

```python
#input()-it is a pre-define function use to take input from user at a run time.
#always take str as an input.
a=input("Enter any number ")
b=input("Enter any number ")
c=a+b
print(c)
print(type(c))
```

```python
#string-Ordered sequence of character or special character.
#it is denoted by '' or " " and a multiline string denoted by single triple quotes ''' '''.
#it is indexed,immutable in nature.
s=""   #empty string
print(s)
print(type(s))
```

```python
s="hello"
print(s)
print(type(s))
print(s[0])   #indexed
s[0]="z"      #immutable
print(s)

```

```python
#list-collection of elements.
#denoted by [] with scv.
#ordered,indexed,mutable and heterogenious in nature.

l=[]   #empty list
print(l)
print(type(l))
```

```python
l=[10,20,30,10,"hello",10.5]
print(l)
print(type(l))
print(l[-1])     #indexed
l[0]=100         #mutable- it does support item assignment
print(l)
```

```python
#Tuple()=collection of elements.
#denoted by () with csv.
(#ordered,indexed,immutable and heterogenous in nature.
t=()
print(t)
print(type(t))
```

```python
#single valued tuple
t=(10,)
print(t)
print(type(t))
```

```python
t=(10,20,30,10,"hello",10.5)
print(t)
print(type(t))
print(t[0])  #indexed
t[0]=100     #immutable
print(t)
```

```python
#Dictionary--Dictionay is a collection of elements
#denoted by curly bracket with key value pairs{k:y},key must be unique but value can be duplicate.
#ordered,unindexed and mutable.
d={}    #Empty dictionary
print(d)
print(type(d))
```

```python
d={"name":"abc","subject":"DA","fees":30000}
print(d)
print(type(d))
print(d["name"])     #unindexed
d["name"]="pqr"      #mutable
print(d)
```

```python
d={"name":"abc","subject":"DA","fees":30000}
print(d)
print(type(d))
d["Name"]="pqr"      
print(d)
```

```python
d={}
print(d)
d["name"]="abc"
print(d)
```

```python
d={"name":["abc","pqr","def"],"subject":["DA","DS","DE"]}
print(d)
print(d["name"][-1])
d["name"][-1]="xyz"
print(d)
```

```python
#SET---set is a collection of unique elements.
#Denoted by {} with CSV
#unordered,unindexed,mutable and hetrogenious in nature.
s=set()      #set()---set constructor
print(s)
print(type(s))
```

```python
s={10,20,10,20,"hello",10.5,10,20,100}
print(s)
print(type(s))
s.add(1000)    #mutable
print(s)
```

```python
#Frozen set--collection of unique elements.
#denoted by a constructor.
#unordered,unindexed,immutable, and hetrogenous in nature.
f=frozenset()
print(type(f))
```

```python
f=frozenset({10,20,10,20,"hello",10.5})
print(f)
print(type(f))
```
