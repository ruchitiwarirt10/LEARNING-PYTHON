```python
#String- It is a sequence of character
#Character-character should be a-z,A-Z,Numeric value,Special symbol,Space.
#it is written in single and double quotes.
#it is indexed,immutable,ordered,heterogeneous.

s="Gagan"
print(s)
print(type(s))
print(s[0])
s[0]="N"    #We can not change value of any index position in string
```

```python
#Slicing:-Slicing is
#string[start:end:step]
s="Decodedataacademy"
print(s[2::])

s="Decodedata"
print(s[2::])
print(s[2:7:])
print(s[:7:])
print(s[::3])
print(s[1:8:2])
print(s[::-1]) #Reverse String

```

```python
a="DataAnalyst"
print(a[2:7:])
print(a[1::2])
print(a[::-1])
```

```python
#Looping On String:-
#1-Direct Iteration
#2-Indirect Iteration

s="DecodeData"
for a in s:             #Direct Iteration
    print(a)

for i in range(len(s)): #Indirect Iteration
    print(s[i])
```

```python
#find length of string without using len()
s="DecodeData"
count=0
for i in s: #dirct iteration
    count=count+1
    print(i,count)

print(count)
print(len(s))
```

```python
#Reverse String Without Slicing:-
s="DecodeData"
rev=""
for i in s:
    rev=i+rev

print(rev)
```

```python
#Operation on string:-
#+,*,membership Operator(in,not in)

s1="Decode "
s2="Data"
print(s1+s2) 
print(s1*3)
print("e" in s1)
print("E" is s1)
print(s1==s2)

# == - it is used to compare values
# is - it is used to compare address
```
