```python
#Tuple: It has collection of comma seperated data in round bracket()
#It is ordered,Immuateble,heterogenious (multiple datatypes),indexed

t=tuple()
print(type(t))  #empty Tuple

a=(10,)
print(type(a))

b=[10,20,30]
print(b)
c=tuple(b)      #Typecasting
print(c)
```

```python
#list=mutable
#tuple=immutable
t=(10,"Gagan",10.5,[100,200])
print(t[0])   #Indexing
print(t[2])
t[2]='DA'     #Error Because Tuple is Immutable

t=[10,"Gagan",10.5,[100,200]]
print(t[0])   #Indexing
print(t[2])
t[2]='DA'   
```

```python
t=(10,"Gagan",10.5,[100,200],55,65)
print(t[1::])   
print(t[:4:])
print(t[::2])   
print(t[::-1])
```

```python
#Looping through Tuple
##Direct Looping
##Indirect Looping(using range())

t=(10,"Gagan",10.5,[100,200],55,65)
for i in t:  #DIrect looping
    print(i,end=" ")
    print()


for i in range(len(t)):  #Indirect Looping
    print(t[i],end=" ")
```

```python
#Tuple methods
#count()-to find occurance of particular element
#index(value,[index])-to find index value of any particular element,2nd arrgument is optional
#Enumerate method-it gives index-value pair,enumerate(tuple,start=0)
t=(10,10,20,20,20,30,40,50)
print(t.count(20))
print(t.index(20,3))

print("----------------------")

for i in enumerate(t,start=5):
    print(i)

print("----------------------")

for i in enumerate(t):
    print(i)
```

```python
#Zip Method--
#zip()-It is used to combine multiple tuple
names=("Ruchi","Bhavya","Alia","Laxmi")
course=("DA","DS","ML","BA")
fees=(25000,40000,45000,30000)
for i in zip(names,course,fees):
    print(i)
```
