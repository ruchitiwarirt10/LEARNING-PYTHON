```python
# check string Palindrome or not
# madam,level,mom,radar
s=input("Enter String")
rev=""
for i in s:
    rev=i+rev
if s==rev:
    print("Palindrome String")
else:
    print("Not Palindrome String")
```

```python
# slicing method
s=input("Enter String")
if s==s[::-1]:
    print("Palindrome String")
else:
    print("Not Palindrome String")
```
