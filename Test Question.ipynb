```python
def prime(start,stop):
    l=[]
    for n in range(start,stop+1):
        if n>1:
            for i in range(2,n):
                if n%i==0:
                    break

            else:
                l.append(n)
    return l
start = int(input("Enter start: "))
stop = int(input("Enter stop: "))

print("Prime numbers:", prime(start, stop))
```

```python
n=int(input("enter any number "))
for i in range(1,n):
    if i==1:
        print(i)
    else:
        print(i,end=" ")
        for j in range(2*i-3):
            print("*",end=" ")
    print(i)
```
