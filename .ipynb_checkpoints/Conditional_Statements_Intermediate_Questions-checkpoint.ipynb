```python

#Conditional_Statements_Intermediate_Questions.txt
#100% PYTHON CONDITIONAL STATEMENTS - INTERMEDIATE CODING QUESTIONS (11-20)

#11. Largest of Three Numbers
#Explanation:
#Find the largest among three numbers.
#Concepts:
#- if...elif...else

a=int(input( "Enter Number a : "))
b=int(input( "Enter Number b : "))
c=int(input( "Enter Number c : "))
if a>b and a>c:
    print(a,"IS GREATER")
elif b>a and b>c:
    print(b,"IS GREATER")
else:
    print(c,"IS GREATER")
```

```python
#12. Grade Calculator
#Explanation:
#Assign grades based on marks.
#Concepts:
#- Multiple conditions

marks=int(input("Enter Marks : "))
if marks>=85:
    print("A Grade")
elif marks>=70 and marks<85:
    print("B Grade")
elif marks>=50 and marks<70:
    print("C Grade")
elif marks>=35 and marks<50:
    print("D Grade")
elif marks<0:
    print("PLEASE ENTER VLID MARKS")
else:
    print("Fail")
```

```python
#13. Leap Year Checker
#Explanation:
#Determine whether a year is leap year.
#Concepts:
#- Nested conditions
#- Modulus

year=int(input("Enter year to check : "))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print("This is leap year")
        else:
            print("This is not leap year")
    else:
        print("leap year")
else:
    print("Year is not leap year")
```

```python
#14. Electricity Bill Calculator
#Explanation:
#Calculate bill according to unit slabs.
#Concepts:
#- if...elif...else
unit=int(input("Enter unit to check : "))
if unit<100:
    print("unit saver")
elif unit>100:
    print("unit consumes as per usage")
else:
    print("hight unit use")
```

```python
#15. Salary Bonus
#Explanation:
#Give 10% bonus if salary > 50000 else 5%.
#Concepts:
#- Arithmetic
#- Conditions

salary=int(input("Enter salary  : "))
if salary>50000:
    salary=salary+salary*10/100
    print("salary after 10 percent bonus : ",salary)
else:
    salary=salary+salary*5/100
    print("salary after 5 percent bonus : ",salary)
```

```python
#18. Triangle Type
#Explanation:
#Identify Equilateral, Isosceles or Scalene triangle.
#Concepts:
#- Comparison
a=int(input("Enter side a"))
b=int(input("Enter side b"))
c=int(input("Enter side c"))
if a==b and a==c and b==c:
      print("equilateral")
elif(a==b and a!=c) or (b==c and b!=a) or (a==c and a!=b):
      print("isosceles")
else:print("scalene")
```
