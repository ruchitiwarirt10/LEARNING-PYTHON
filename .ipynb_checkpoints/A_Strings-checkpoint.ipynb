## Index/Indexing--

```python
#index-Position of a single character in a given string.
#first character of any string is at 0 index.
#last charcacter is at -1 index.

#indexing- it is a technique for accessing individaul charcater from a given string

s="welcome to bhopal"
print(s[0])
print(s[-1])
print(s[7])
print(s[-4])
```

## Slicing--

```python
#slicing-it is a technique for extracting substing from a given string.
#strobj[start:stop:step]
#start-by default-0 index
#stop-by default end of the string or n-1
#step-by default- +1 increment
s="Welcome to bhopal"
print(s[::])
print(s[:7:])
print(s[::-1])
print(s[-1:-7:-1])
```

```python
s="welcome to city of lakes"
print(s[1:15:2])
print(s[-2:-16:-2])
print(s[:11:])
print(s[0:7:2])
print(s[-1:-14:-2])
print(s[1: :3])
```

## Iteration--

```python
#iteration
#direct iteration
s="hello"
for i in s:
    print(i)
```

```python
#indirect iteration-with help of range function
#len()-no of charcater in a given string.
s="hello"
for i in range(len(s)):
    print(i,s[i])
```

```python
#wap to print no of vowels no of consonents and the no. of spaces in a given string?
s="welcome to city of lakes"
c_vowel=0
c_con=0
c_s=0
for i in s:
    if i in "aeiouAEIOU":
        c_vowel=c_vowel+1
    elif i==' ':
        c_s=c_s+1
    else:
        c_con=c_con+1
print("count of vowel: ",c_vowel)
print("count of consonent: ",c_con)
print("count of spaces: ",c_s)
        

    
```

```python
#wap to print reverse of an any given string.?
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
print(rev)         #without slicing
print(s[::-1])     #with slicing

      
```

```python
#wap to check whether a given string is palindrome or not.?
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
if s==rev:
    print(s,"is a palindrome")
else:
    print(s,"is not a palindrome")
```

```python
s=input("enter any string: ")
rev=''
if s==s[::-1]:
    print(s,"is a palindrome")
else:
    print(s,"is not a palindrome")
```

## String Functions--

```python
#Function
#len()-return no of character in a given string.
#len(strobj)
s="hello"
print(len(s))
count=0
for i in s:
    count=count+1
print(count)
```

```python
#chr(number)=takes number(ascii) as an input and return character.
print(chr(65))
print(chr(97))
print(chr(32))
print(chr(90))
print(chr(122))
```

```python
#ord(character)-take character or special character as an input and return number.
print(ord('A'))
print(ord('a'))
print(ord(' '))
```

```python
s="bhopal"
max_value=s[0]
for i in s:
    if i>max_value:
        max_value=i
print(max_value)
```

```python
s="bhopal"
min_value=s[0]
for i in s:
    if i<min_value:
        min_value=i
print(min_value)
```

```python
n=int(input("Enter number: "))
for i in range(n):
      for j in range(i+1):
          print("*",end=" ")
      print()
```

```python
n=5
for i in range(n):
    for j in range(i+1):
        print(chr(65+i),end=" ")
    print()
```

```python
n=5
for i in range(n):
    for j in range(i+1):
        print(chr(65+j),end=" ")
    print()
```

```python
n=5
for i in range(n):
    for j in range(n-1-i):
        print(" ",end=" ")
    for j in range(i+1):
        print(chr(65+j),end=" ")
    print()
```

```python
n=5
for i in range(n):
    for j in range(n-i-1):
        print(" ",end=" ")
    for j in range(i+1):
        print(chr(65+j),end=" ")
    for j in range(i):
        print(chr(65+j),end=" ")
    print()
```

## Methods--

### upper(),lower(),title().capitalize()

```python
s="welcome to city of lakes"
print(s.upper())
print(s.lower())
print(s.title())
print(s.capitalize())
```

```python
s="welcome to city of lakes"
print(s.isupper())
print(s.islower())
print(s.istitle())
```

```python
s="hello"
s1="hello123"
print(s.isalpha())
print(s.isalnum())
print(s.isdigit())
```

```python
#count():-return occurrence of any character in a given string.
#strobj.count(chr/sustring,[start],[stop])
#start- by default-0
#stop- by default-end of the string
#stop-n----->(n-1)
s="welcome to bhopal"
print(s.count('o'))
print(s.count('o',5))
print(s.count(' '))
print(s.count('bhopal'))
print(s.count('z'))
print(s.count('o',5,9))
```

```python
#split()
#strobj.split([seprator],[maxsplit])
#seprator- by default space
#maxsplit--max. occurance seprator
#always return a list of string
s="welcome to city of lakes"
l=s.split()
print(l)
```

```python
s="welcome to city of lakes"
l=s.split(' ',1)
print(l)
```

```python
s="welcome to city of lakes"
l=s.split('o')
print(l)
```

```python
#join
#separator.join(string)
l=["welcome","to","bhopal"]
s=" ".join(l)
print(s)
```

```python
s="hello"
l="/".join(s)
print(l)
j=l.split("/")
print(j)
```

```python
#replace
#strobj.replace(oldstring,newstring/newcharacter,[count])
s="welcome to bhopal"
s1=s.replace("bhopal","decode")
print(s1)
```

```python
s="welcome to city of lakes"
s1=s.replace('o','a')
print(s1)
s2=s.replace('o','a',1)
print(s2)
```

```python
#Split--remove unwanted leading and trailing character or special character from given string.
#strobj.strip([character])
#by default-spaces
s=' welcome  to city of lakes          '
s1=s.strip()
print(s1)
print(len(s))
print(len(s1))
```
