```python
# 1. Find the Largest of Three Numbers
# Write a Python program that accepts three numbers from the user and displays the largest number among them.
# INPUT REQUIRED: Input required: Three numbers.
# EXPECTED OUTPUT: Expected output: The largest number.
a=int(input("enter any number: "))
b=int(input("enter any number: "))
c=int(input("enter any number: "))
if a>b and a>c:
    print("a is largest")
elif b>a and b>c:
    print("b is largest")
else:
    print("c is largest")
    
```

```python
# 2. Check Whether a Year Is a Leap Year
# Write a Python program that accepts a year from the user and checks whether it is a leap year. A year is a leap year if it
# is divisible by 400, or if it is divisible by 4 but not by 100.
# INPUT REQUIRED: Input required: A year (integer).
# EXPECTED OUTPUT: Expected output: Display whether the year is a leap year or not.
year=int(input("enter any number: "))
if year%4==0 and year%100!=0 or year%400==0:
    print(year,"is a leap year")
else:
    print(year,"is not a leap year")
```

```python
# 3. Check Whether a Number Is a Multiple of Both 3 and 7
# Write a Python program that accepts an integer and checks whether it is exactly divisible by both 3 and 7.
# INPUT REQUIRED: Input required: One integer.
# EXPECTED OUTPUT: Expected output: Display whether the number is divisible by both 3 and 7.
num=int(input("Enter number: "))
if num%3==0 and num%7==0:
    print(num,"is divisible by both")
else:
    print(num,"is not divisible by both")
```

```python
# 4. Student Grade Calculator
# Write a Python program that accepts a student’s marks out of 100 and assigns a grade according to the rules below.
# First, check whether the marks are valid (between 0 and 100).
# INPUT REQUIRED: Input required: Student’s marks.
# EXPECTED OUTPUT: Expected output: The student’s grade or an invalid-marks message.
# Marks    Grade
# 90–100    A+
# 80–89     A
# 70–79     B
# 60–69     C
# 40–59     D
#Below 40 Fail
Marks=int(input("enter marks: "))
if Marks>=0 and Marks<=100:
    if Marks>=90:
        print(Marks,"grade A+")
    elif Marks>=80:
        print(Marks,"grade A")
    elif Marks>=70:
        print(Marks,"grade B")
    elif Marks>=60:
        print(Marks,"grade C")
    elif Marks>=40:
        print(Marks,"grade D")
    else:
        print("Fail")
else:
    print(Marks,"Invalid")
```

```python
# 5. Student Scholarship Eligibility
# Write a Python program that accepts a student’s marks and attendance percentage. The student is eligible for a
# scholarship only if marks are greater than or equal to 85 AND attendance is greater than or equal to 75%.
# INPUT REQUIRED: Input required: Marks and attendance percentage.
# EXPECTED OUTPUT: Expected output: Display whether the student is eligible for the scholarship.
Marks=int(input("Enter Marks: "))
Attendance=int(input("Enter attendence: "))
if Marks>=0 and Marks<=100 and Attendance>=0 and Attendance<=100:
    if Marks>=85 and Attendance>=75:
        print("Student is eligible for scolership")
    else:
        print("Student is not eligible for scolership")
else:
    print("invalid Marks and attendance")
```

```python
# 6. Salary Bonus Calculator
# Write a Python program that accepts an employee’s monthly salary. If the salary is greater than ₹50,000, the
# employee receives a 10% bonus; otherwise, the employee receives a 5% bonus. Calculate the bonus and total salary
# after adding the bonus.
# INPUT REQUIRED: Input required: Monthly salary.
# EXPECTED OUTPUT: Expected output: Bonus amount and total salary after bonus.
salary=float(input("Enter Salary: "))
if salary>50000:
    bonus=(salary*10)/100
else:
    bonus=(salary*5)/100
salary=salary+bonus
print("bonus amount ",bonus)
print("total salary",salary)        
```

```python
# 7. Electricity Bill Calculator
# Write a Python program that accepts the total electricity units consumed and calculates the bill using progressive
# slab rates. 
#First 100 units: ₹5 per unit; 
#next 100 units (101–200): ₹7 per unit; 
#above 200 units: ₹10 per unit. For
# example, for 250 units, charge 100 units at ₹5, the next 100 at ₹7, and the remaining 50 at ₹10. Reject negative unit
# consumption. Assume there are no fixed charges or additional taxes.
# INPUT REQUIRED: Input required: Total units consumed.
# EXPECTED OUTPUT: Expected output: Total electricity bill in rupees.
# #:-Solution-----
unit=int(input("Enter unit: "))
if unit<0:
    print("negative unit")
elif unit<=100:
    per_unit=unit*5
    print("total_electricity",per_unit)
elif unit<=200:
    per_unit=unit*7
    print("total_electricity",per_unit)
else:
    per_unit=(100*5)+(100*7)+((unit-200)*10)
    print("total electricity",per_unit)    
```

```python
# 9. Simple Calculator
# Write a Python program that accepts two numbers and an arithmetic operator from the user. Perform the calculation
# for +, -, * or /. Display an appropriate message for an unsupported operator and prevent division by zero.
# INPUT REQUIRED: Input required: Two numbers and one operator.
# EXPECTED OUTPUT: Expected output: The calculated result or an appropriate error message.
a=int(input("Enter any number: "))
b=int(input("Enter any number: "))
op=input("+,-,*,/")
if op=="+":
    print(a+b)
elif op=="-":
    print(a-b)
elif op=="*":
    print(a*b)
elif op=="/":
    if b!=0:
        print(a/b)
    else:
        print("Error")
else:
    print("invalid operator")
```

```python
# 10. Check Whether a Character Is an Alphabet
# Write a Python program that accepts a single character and checks whether it is an English alphabet letter. Reject
# digits, spaces and special characters. Also check that the user entered exactly one character.
# INPUT REQUIRED: Input required: One character.
# EXPECTED OUTPUT: Expected output: Display whether the character is an alphabet or not.
ch=input("enter character: ")
if len(ch)==1:
    if (ch>='A' and ch<='Z') or (ch>='a' and ch<='z'):
        print(ch,"is an alphabat")
    else:
        print(ch,"is not an alphabat")
else:
    print("Enter only one character")
```

```python
# 11. Vowel or Consonant
# Write a Python program that accepts a single character and determines whether it is a vowel or a consonant. Accept
# uppercase and lowercase letters. If the input is not a single alphabet letter, display an appropriate message.
# INPUT REQUIRED: Input required: One character.
# EXPECTED OUTPUT: Expected output: “Vowel”, “Consonant” or “Invalid Input”
ch=input("Enter character: ")
if len(ch)==1:
    if (ch>='A' and ch<='Z') or (ch>='a' and ch<='z'):
        if ch in "aeiouAEIOU":
            print(ch,'is Vowel')
        else:
            print(ch,'is consonant')
    else:
        print("invalid input")
else:
    print("enter only one character")
```

```python
# 12. ATM Withdrawal System
# Write a Python program that accepts the user’s current bank balance and the amount they want to withdraw. The
# withdrawal amount must be greater than zero and must not exceed the available balance. If valid, deduct the
# amount; otherwise, display an appropriate message.
# INPUT REQUIRED: Input required: Current balance and withdrawal amount.
# EXPECTED OUTPUT: Expected output: Transaction status and remaining balance if successful.

current_balance=float(input("Enter Amount: "))
withdrawal_Amount=float(input("Enter withdrawal Amount: "))
if withdrawal_Amount>0 and withdrawal_Amount<=current_balance:
    Remaining_balance=current_balance-withdrawal_Amount
    print("Transaction Successful ")
    print("Remaining balance:",Remaining_balance)
else:
    print( "withdrawal amount must be greater than zero and less or equal to current balance")                 
```
