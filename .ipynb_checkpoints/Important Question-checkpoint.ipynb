```python
# check Armstrong Number
num=int(input("Enter Number to check : "))
s=0
temp=num
power_of_digit=len(str(num))
while temp!=0:
    digit=temp%10
    s=s+digit**power_of_digit
    temp=temp//10

if s==num:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
num=int(input("Enter Number to check : "))
s=0
temp=num
power_of_digit=len(str(num))

for i in range(len(str(num))):

    digit=temp%10    # It gives last digit
    s=s+digit**power_of_digit    # It gives sum of power of digit
    temp=temp//10  # It gives floor value

if s==num:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
            
```

```python
# Check prime number
num=int(input("Enter Number : "))   #10
if num==0 or num==1:
    print("Number is not prime")
else:
    is_prime=True   # Signal or helper variable
    for i in range(2,num):   # (2,9)
        if num%i==0:
            is_prime=False
            break
if is_prime:     # True
    print("Prime Number")
else:
    print("Not Prime Number")
```

```python
# Febonacci series
r=int(input("Enter Term/Range : "))
if r<=0:
    print("Enter Range Greater Than 0")
else:
    a=0
    b=1
    for i in range(r):
        c=a+b
        print(a,end=" ")
        a=b
        b=c
```

```python
# Febonacci series
term=int(input("Enter range : "))  #10
if term<=0:
    print("Enter range greater than 0")
else:
    a=0
    b=1
    for i in range(term):
        print(a,end=" ")
        c=a+b
        a=b
        b=c
```

```python
# Factorial of given number
number=int(input("Enter number : "))
fact=1
for i in range(1,number+1):   # forward
    fact=fact*i
print(fact)

fact1=1
for j in range(number,0,-1):   # reverse
    fact1=fact1*j
print(fact1)
```

```python
# find factorial of number
# Reverse loop
n=5
mul=1
for i in range(n,0,-1):
    mul=mul*i
print(mul)
```

```python
# Forward loop
n=5
mul=1
for i in range(1,n+1):
    mul=mul*i
print(mul)
```

```python
for i in range(1,6):
    for j in range(1,6):
        if j>=6-i:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
# Square
for i in range(1,6):
    for j in range(1,6):
        if i==1 or i==5 or j==1 or j==5:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
# check string Palindrome or not
# madam,level,mom,radar
s=input("Enter String")
rev=""
for i in s:
    rev=i+rev
if s==rev:
    print("Palindrome String")
else:
    print("Not Palindrome String")
```

```python
# greatest among 3 numbers
a=int(input("Enter number a : "))
b=int(input("Enter number b : "))
c=int(input("Enter number c : "))
if a>b and a>c:
    print(a,"is greatest")
elif b>a and b>c:
    print(b,"is greatest")
else:
    print(c,"is greatest")
```

```python
# leap year
## year should be divisible by 400 or year divisible by 4 and not divisible by 100
year=int(input("Enter year to check :"))
if year % 4 == 0:
    if year % 100 ==0:
        if year % 400 ==0:
            print("This is leap year")
        else:
            print("Not leap year")
    else:
        print("leap year")
else:
    print("Year is not leap year")
```

```python
## leap year 
## another method to check leap year or not 

year=int(input("Enter year to check :"))
if year%4==0 and year%100!=0 or year%400==0:
    print("Year is leap year")
else:
    print("Not leap year")
```
