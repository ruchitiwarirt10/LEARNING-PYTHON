```python
#range function
##range (start,stop,step)
#we have to pass atleast one argument which is stop
#for i in range(5):
#        print(i)

#
for i in range(0,5):
    print(i)

#stop
for i in range(1,5):
    print(i)


for i in range(10,0,-1):
    print(i)
```

```python
#step
for i in range(1,10,2):
    print(i)
```

```python
#reverse
for i in range(10,0,-2):
    print(i)
```

```python
#print square of 1 to 10 using range function
for i in range(1, 11):
    print(i**2)        #or (i*i)
```

```python
#print cube of 1 to 10 using range function
for i in range(1,11):
    print(i**3)
```

```python
#type casting
#tuple to list...pass tuple in list() as argument

a=[10,"Gagan",10.5]  #list to tuple
t=tuple(a)
print(t)
print(type(t))

#tuple to list()

a=(10,"ruchi",10.5)   #tuple to list
l=list(a)
print(l)
print(type(l))
```

```python
#type casting
#list to set
#len()

#change it to set
#print that set and their datatype
a=[10,10,20,25.5,20,"Gagan","gagan"]

s=set(a)
print(s)
print(type(s))
print(id(s))
print(bin(2))
print(len(a))
```
