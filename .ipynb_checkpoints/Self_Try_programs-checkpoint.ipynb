#### 1.Palindrom Or not--

##### -Without slicing-

```python
s=input("Enter any string: ")
rev=''
for i in s:
    rev=i+rev
if s==rev:
    print(s,"is a Palindrome")
else:
    print(s,"is not a Palindrome")
```

##### -With Slicing--

```python
s=input("Enter any string: ")
rev=''
if s==s[::-1]:
    print(s,"is a palindrome")
else:
    print(s,"is not a palindrome")
    
```
