```python
num=int(input("ENTER NUMBER  TO CHECK : "))
s=0
temp=num
Power_of_digit=len(str(num))
while temp!=0:
    digit=temp%10             #it gives last digit
    s=s+digit**Power_of_digit #it gives sum of power of digit
    temp=temp//10             #it gives floor value

if s==sum:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```
