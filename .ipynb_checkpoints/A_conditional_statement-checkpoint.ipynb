```python
sub1=int(input("Enter marks of sub1: "))
sub2=int(input("Enter marks of sub2: "))
sub3=int(input("Enter marks of sub3: "))
sub4=int(input("Enter marks of sub4: "))
sub5=int(input("Enter marks of sub5: "))
obt_marks=sub1+sub2+sub3+sub4+sub5
print("obtained marks: ",obt_marks)
per=(obt_marks/500)*100
print("percentage: ",per,"%")
if per>=65:
    print("First Division")
elif per>=50:
    print("Second Division")
elif per>=33:
    print("Third Divison")
else:
    print("Fail")
```

```python
#Q17.Eligibility for special programs
#Objective write a python programs that determines whether a student is eligible for an honors program or needs Remedial classes
#based on their performance in different subjects
#Defaults
#Eligiblity Criteria:-
#Honors:Average Score of 85 or above across all subjects and no
#subject below 70.
#Remedial: Any subject below 50.
#input:
#scores in three Subjects:Math,Science and language(out of 100)
#Output:
#"Eligible for Honors"
#Needs Remedial"
#"Standard placement".
Math=int(input("Enter Math's Marks: "))
Science=int(input("Enter Science's Marks: "))
Language=int(input("Enter Language's Marks: "))
Avg=(Math+Science+Language)/3
if Avg>=85 and Math>=70 and Science>=70 and language>=70:
    print("Eligible for Honors")
elif Math<50 or Science<50 or Language<50:
    print("Needs Remedial")
else:
    print("Standard Placements")
```

```python
#Wap to check Whether A given Year is Leap year or Not?
Year=int(input("Enter Year: "))
if Year%4==0:
    if Year%100==0:
        if Year%400==0:
            print(Year,"is Leap Year")   #Century
        else:
            print(Year,"is Not A leap Year")   #Century
    else:
        print(Year,"is leap year")             #Non-Century
else:
    print(Year,"is Not a Leap Year")           #Non-Century
```

```python
# Q17. Write a Python program that takes a year as input and
# checks if it is a leap year. If it is a leap year, print "Leap
# year," otherwise print "Not a leap year."

year=int(input("Enter Year: "))
if year%4==0 and year%100!=0 or year%400==0:
    print(year,"is a leap year")
else:
    print(year,"is not a leap year")
```

```python
# Q18.Write a Python program that takes an integer as input and
# determines whether it is positive, negative, or zero. Additionally, if
# the number is positive, check if it is even or odd.
num=int(input("Enter any number: "))
if num>0 and num%2==0:
    print(num,"is positive and even")
elif num>0 and num%2!=0:
    print(num,"is positive but odd")
elif num<0:
    print(num,"is negative")
else:
    print(num,"is zero")
```

```python
# Q19. Eligibility for Special Programs

# Objective: Write a Python program that determines whether a
# student is eligible for an honors program or needs remedial classes
# based on their performance in different subjects.

# Details:

# Eligibility criteria:
# Honors: Average score of 85 or above across all subjects and no
# subject below 70.
# Remedial: Any subject below 50.

# Input:
# Scores in three subjects: Math, Science, and Language (out of 100)

# Output:
# "Eligible for Honors"
# "Needs Remedial"
# "Standard Placement"

Math=int(input("Enter Math's Marks: "))
Science=int(input("Enter Science's Marks: "))
language=int(input("Enter language's Marks: "))
Avg=(Math+Science+language)/3
if Avg>=85 and Math>=70 and Science>=70 and language>=70:
    print("eligible for Honors")
elif math<50 or science<50 or language<50:
    print("Needs Remedial")
else:
    print("Standard Placement")
```
