```python
#1. Positive, Negative, or Zero
#Explanation:
#Accept a number and determine whether it is positive, negative, or zero.
#Concepts:
#- if, elif, else
#- Comparison operators

number=int(input("Enter any number : "))
if number>0:
    print("positive")
elif number==0:
    print("zero")
else:
    print("negative")
```

```python
#2. Even or Odd
#Explanation:
#Check whether a number is divisible by 2.
#Concepts:
#- Modulus (%)
#- if...else

number=int(input("Enter any Number : "))
if number%2==0:
    print("Even")

else:
    print("odd")
```

```python
#3. Eligible to Vote
#Explanation:
#Accept age and display whether the person can vote (18 or above).
#Concepts:
#- Comparison operators

Age=int(input("Enter Age : "))
if Age>=18:
    print("eligible for vote")
else:
    print("Not eligible for vote")
 
```

```python
#4. Largest of Two Numbers
#Explanation:
#Compare two numbers and display the larger one.
#Concepts:
#- if...else

Number=int(input("Enter any Number : "))
if Number>50:
    print("largest number")
else:
    print("smallest")
```

```python
#5. Divisible by 5
#Explanation:
#Check whether a number is divisible by 5.
#Concepts:
#- Modulus
#- if

number=int(input("Enter any number : "))

if number%5==0:
        print("divisible by 5")
```

```python
#6. Pass or Fail
#Explanation:
#Display Pass if marks are 35 or above, otherwise Fail.
#Concepts:
#- Comparison

marks=int(input("Enter any Number : "))
if marks>=35:
    print("pass")
else:
    print("Fail")
```

```python
#7. Check Alphabet
#Explanation:
#Accept a character and determine whether it is an alphabet.
#Concepts:
#- Character comparison

character=str(input("Enter any character : "))
if character .isalpha():
    print(f"'{character}' is an alphabet.")
else:
    print(f"'{character}'is not an alphabet.")
```

```python
#8. Vowel or Consonant
#Explanation:
#Determine whether the entered alphabet is a vowel or consonant.
#Concepts:
#- in operator

letter=str(input("Enter an alphabet letter: "))
if letter in 'aeiou':
    print(f"'{letter}' is a vowel.")
else:
    print(f"'{letter}' is consonent.")
```

```python
#9. Number Sign
#Explanation:
#Display Positive, Negative or Zero.
#Concepts:
#- if...elif...else
number=int(input("Enter any number : "))
if number>=0 and number<=10:
           print("zero")
elif number>=10:
             print("positive")
else:
     print("negative")
```

```python
#10. Multiple of 3 and 7
#Explanation:
#Check whether a number is divisible by both 3 and 7.
#Concepts:
#- and
#- Modulus

number=int(input("Enter any number : "))
if number%3==0 and number%7==0:
    print("multiple of 3 and 7")
else:
    print("not multiple of 3 and 7")
```

```python
#LEAP YEAR CHECK CONDITIONS-
#(YEAR%400==0)OR(YEAR%4==0 AND YEAR%100!=0)
#YEAR SHOULD DIVISIBLE BY 400 OR YEAR DIVISIBLE BY 4 AND NOT DIVISIBLE BY 100
year=int(input("Enter year to check : "))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print("This is leap year")
        else:
            print("This is not leap year")
    else:
        print("leap year")
else:
    print("Year is not leap year")
```

```python
year=int(input("Enter year to check : "))
if year%400==0 or year%4==0 and year%100!=0:
    print("year is leap year")
else:
    print("not leap year")
```

```python
#factorial of given number-5!=5*4*3*2*1,4!=4*3*2*1,n!=n*(n-1),*(n-2)*.......*1
number=int(input("Enter a number : "))   #Forward
fact=1
for i in range(1,number+1):
    fact=fact*i
print(fact)
```

```python
number=int(input("Enter a number : "))   #Reverse
fact1=1
for j in range(number,0,-1):
    fact1=fact1*j
print(fact1)
```
