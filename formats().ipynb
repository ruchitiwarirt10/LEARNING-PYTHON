```python
#Q3.
password=input("Enter Password : ")
is_upper=False    #helper variable or signal
is_digit=False     #helper variable or signal
for i in password:  #direct looping
    if i.isdigit():
        is_digit=True
    elif i.isupper():
        is_upper=True

if len(password)>8 and is_digit and is_upper:  #true,true,true
    print("valid Password")
else:
    print("Invalid Password")
```

```python
#Q4. vowels and consonant count in string-
s=input("Enter String : ")
s=s.lower()
vowel_count=0
consonant_count=0
for i in s:
    if i in 'aeiou':
        vowel_count=vowel_count+1
    elif i not in 'aeiou' and i!=" ":
        consonant_count+=1
print(vowel_count)
print(consonant_count)
```
