```python
"""PYTHON FILTER() + LAMBDA
============================================================
QUESTION 1: FILTER EVEN NUMBERS
============================================================
Question:
Extract only even numbers.

Code: numbers = [10, 15, 22, 31, 40, 51, 60]

Expected output:[10, 22, 40, 60]

Hint:x % 2 == 0"""

a=[10, 15, 22, 31, 40, 51, 60]
numbers=filter(lambda x:x%2==0,a)
print(list(numbers))
```

```python
"""============================================================
QUESTION 2: FILTER ODD NUMBERS
============================================================

Question:
Extract only odd numbers.

Code:

    numbers = [11, 20, 33, 44, 55, 68, 71]

Expected output:

    [11, 33, 55, 71]"""

a=[11, 20, 33, 44, 55, 68, 71]
number=filter(lambda x:x%2!=0,a)
print(list(number))
```
