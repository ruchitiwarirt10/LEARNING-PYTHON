```python
#7.  Check Prime Number
#-   Write a function to check whether a number is prime.

def prime(num):
    if num<1:
        return "Not prime Number"
    else:
        is_prime=True
        for i in range(2,num):
            if num%i==0:
                is_prime=False
                break
        if is_prime:
            return "Prime Number"
        else:
            return "Not prime Number"
print(prime(12))
```

```python
#8.  Print Prime Numbers in a Given Range
#-   Write a function to print list of all prime numbers between two given numbers.
def prime_list(start,stop):
    l=[]
    if start>stop or start<0 or stop<0:
        return "Enter Valid Range"
    else:
        for i in range(start,stop+1): #start=10,stop=100
            is_prime=True
            for j in range(2,i):      #2-10
                if i%j==0:
                     is_prime=False
                     break
            if is_prime:
                l.append(i)
    return l
print(prime_list(10,100))
```

```python
#9.  Generate Fibonacci Series
#-   Write a function to generate the list of Fibonacci series up to N terms.
def feb(num):   #num=5    #o/p=0 1 1 2 3
    a=0
    b=1
    if num<=0:
        return "Enter Positive Integer"
    else:
        if num==1:
            return [a]
        elif num==2:
            return [a,b]
        else:
            feb_list=[a,b]
            for i in range(3,num+1):   #num=3
                c=a+b                  #[0,1,1]
                feb_list.append(c)
                a=b
                b=c
    return feb_list
print(feb(5))
```
