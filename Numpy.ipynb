```python
#broadcasting in numpy is a powerful technique that allows mathemetical operation between 2 arrays of different shape and dimension.
#broadcasting Rule--

#When operating 2 arrays numpy evaluate there shape element wise starting from right most to the left most.

#if 2 dimension array are compatible

#1.they are same, 

#2.if dimensions  are  different then one of them is exactly one,if two arrays of diffrent dimension,the shape of smaller array 
#is left paded with ones.
#-example-arr1-->shape(3,)-->1D,
#arr2-->shape(2,3)-->2D

#3.Compatibility Error rule--if any dimension array size disagree and nither size is = to 1 broadcasting fails.and raise a value error.
```

```python
import numpy as np
```

```python
#1-D Array--
arr1= np.array([10,20,30,40])
arr2=np.array([100,200,300,400])
print(arr1)
print(arr2)
print("-----*****------")
arr_new=np.stack((arr1,arr2))
print(arr_new)
print("-----*****------")
arr_new=np.stack((arr1,arr2),axis=0)
print(arr_new)
print("-----*****------")
arr_new=np.stack((arr1,arr2),axis=1)
print(arr_new)
```

```python
#2D ----
#2D-Array
arr1=np.array([[100,200,300,400],[500,600,700,800]])
arr2=np.array([[10,20,30,40],[50,60,70,80]])
print(arr1)
print(arr2)
print("-----*****------")
arr_new=np.stack((arr1,arr2))
print(arr_new)
print("-----*****------")
arr_new=np.stack((arr1,arr2),axis=1)
print(arr_new)
print("-----*****------")
arr_new=np.stack((arr1,arr2),axis=2)
print(arr_new)
```
