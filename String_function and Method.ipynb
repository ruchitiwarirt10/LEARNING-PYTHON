```python
#Find factorial of number
n=int(input("Enter Number: "))
mul=1
for i in range(n,0,-1):   #Reverse loop
    mul=mul*i
print(mul)
```

```python
n=int(input("Enter Number: "))
mul=1
for i in range(1,n+1):   #Forward loop
    mul=mul*i
print(mul)
```
