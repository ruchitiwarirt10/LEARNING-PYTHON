```python
#Ascending Order
l=[10,7,100,40]
#[7,10,40,100]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        if l[i]>l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```

```python
#Desending Order-
l=[10,7,100,40]
#[7,10,40,100]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        if l[i]<l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```
