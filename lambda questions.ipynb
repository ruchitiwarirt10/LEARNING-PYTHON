```python
#PYTHON LAMBDA FUNCTION

#QUESTION 1: ADD TWO NUMBERS
#---------------------------
#Create a lambda function that takes two numbers and returns their sum.

result= lambda a,b:a+b
print(result(10,10))
```

```python


#QUESTION 2: SUBTRACT TWO NUMBERS
#--------------------------------
#Create a lambda function that takes two numbers and returns the result
#of the first number minus the second number.

substraction= lambda a,b:a-b
print(substraction(30,10))


```

```python
#QUESTION 3: MULTIPLY TWO NUMBERS
#-------------------------------
#Create a lambda function that takes two numbers and returns their
#multiplication.

multiply= lambda a,b:a*b
print(multiply(20,20))


#QUESTION 4: DIVIDE TWO NUMBERS
#------------------------------
#Create a lambda function that takes two numbers and returns their
#division.

divide= lambda a,b:a/b
print(divide(10,5))


#QUESTION 5: FIND THE SQUARE OF A NUMBER
#---------------------------------------
#Create a lambda function that accepts one number and returns its square.

square= lambda x:x*x
print(square(10))


#QUESTION 6: FIND THE CUBE OF A NUMBER
#-------------------------------------
#Create a lambda function that accepts one number and returns its cube.

cube= lambda a:a*a*a
print(cube(20))


#QUESTION 7: FIND THE REMAINDER
#------------------------------
#Create a lambda function that accepts two numbers and returns the
#remainder when the first number is divided by the second number.

reminder= lambda a,b:a%b
print(reminder(10,3))
```

```python
#QUESTION 8: FIND THE QUOTIENT
#-----------------------------
#Create a lambda function that accepts two integers and returns the
#integer quotient of their division.

quotient= lambda a,b:a//b
print(quotient(20,3))
```

```python
#QUESTION 9: ADD 10 TO A NUMBER
#------------------------------
#Create a lambda function that accepts a number and adds 10 to it.
num= lambda a:a+10
a=int(input("Enter any number"))
print(num(a))
```

```python
#QUESTION 10: CONVERT CELSIUS TO FAHRENHEIT
#-------------------------------------------
#Create a lambda function to convert temperature from Celsius to Fahrenheit.
#F = °C * 9/5 + 32
fer= lambda cel:cel*9/5+32
print(fer(32))
```

```python
#QUESTION 11: CALCULATE SIMPLE INTEREST
#--------------------------------------
#Create a lambda function to calculate Simple Interest.

SI= lambda p,r,t:p*r*t/100
print(SI(1000,10,2))
```

```python
#lambda function with map()
#map()-when we want tonapply function to every element of Iterables (list,tuple,set) then we use map()>
#it is weitten in sigle expression
#syntax : map(function,iterables)
a=[2,4,6,8]
square= map(lambda x:x**2,a)
print(list(square))
```
