```python
##OOps-object oriented programming system/structure

##It is a programming technique in which we create programs using class and object.
###Class-It is a blueprint or template.
##Object-It is an instancese of class

##We create class using keyword class.

class House:
    pass
a1=House()
a2=House()
print(a1)
print(a2)
```

```python
##Self=it stores object data nsd handle particular object's data
class person:
    def __init__(self,name,age):          #Constroctor method
        self.name=name
        self.age=age
    
p1=person("gagan",32)
print(p1.age)
print(p1.name)
```

```python
class Account:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
account1=Account("Gagan",50000)
print(account1.name)
print(account1.balance)
account1.balance=-5000
print(account1.balance)
```

```python
##Pillers of OOps-
#1.Encapsualtion-Wrapping data and method in single unit(class) and restrict direct access of object's data outside of class.
##a. Access specifier-provide security to object data
###1.public Variable-self.name=name
###2.Protected variable-Self._name=name
###3.private variable-self.__name=name
##Getter method and setter method-getter 
#and setter method used to set or get private variable or used to perform operations on private variable

#Decorators-Decorators is a function that takes another function and adds extra behaviour to it without 
#changing the original function code(function that ).

#2.Inheritance
#3.Ploymorphism
#4.Abstraction

class Account:
    def __init__(self,name,balance):
        self.name=name                   #Public variable
        self.__balance=balance           #Private Variable

    def getbalance(self):
        print("Balance is ",self.__balance)

    def setbalance(self,amount):
        self.__balance=amount
a1=Account("Gagan",5000)
print(a1.name)
a1.getbalance()
a1.setbalance(20000)
a1.getbalance()
```

```python
#Method 1 of getter and setter

class Account:
    def __init__(self,name):
     self.__name=name

    def get_name(self):      #Traditional getter method
        print(self.__name)
        
    def set_name(self,newname):    #Traditional setter method
        self.__name=newname

a=Account("Gagan")
a.get_name()
a.set_name("Ruchi")
a.get_name()
```

```python
#Method 2 of getter and setter
#we use @property decorator

class Account:
    def __init__(self,name):
     self.__name=name

    @property
    def name(self):      #getter method
        print(self.__name)
        
    @name.setter  
    def name(self,newname):    #setter method
        self.__name=newname

a=Account("Gagan")
print(a.name)
a.name="Ruchi"
print(a.name)

```
