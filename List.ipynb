```python
#Print Prime Number in Given Range-
start=int(input("Enter start : "))
end=int(input("Enter End : "))
prime_list=[]
if start<0 or end<0 or start>end:
    print("print valid range")
else:
    for i in range(start,end+1):  #start=10 End=100
        is_prime=True
        for j in range(2,i): #2-9
            if i%j==0:
                is_prime=False
                break
        if is_prime:
            prime_list.append(i)
print(prime_list)
```
