```python

#Python_Map_with_Lambda_function.txt
#============================================================
#QUESTION 1: DOUBLE EVERY NUMBER
#============================================================
#Question:
#Create a map() + lambda expression that doubles every number.
#Code:
#numbers = [1, 2, 3, 4, 5]
#Expected output:[2, 4, 6, 8, 10]
a=[1, 2, 3, 4, 5]
double=map(lambda x:x*2,a)
print(list(double))
```
