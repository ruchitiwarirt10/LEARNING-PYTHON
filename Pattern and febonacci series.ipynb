```python
for i in range(1,6):
    for j in range(1,6):
        if j>=6-i:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        if i==1 or i==5 or j==1 or j==5:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
#Swap Number without using third variable(method-1)
a=10
b=20
print(f"Number before Swapping a is {a} and b is {b}")
a=a+b
b=a-b
a=a-b
print(f"Number after Swapping a is {a} and b is {b}")
```

```python
#Swap Number without using third variable(method-2)
a=10
b=20
print(f"Number before Swapping a is {a} and b is {b}")
a,b=b,a #tuple unpaking
print(f"Number after Swapping a is {a} and b is {b}")
```

```python
#Swap Number with using third variable(method-3)
a=10
b=20
print(f"Number before Swapping a is {a} and b is {b}")
c=a
a=b
b=c
print(f"Number after Swapping a is {a} and b is {b}")
```

```python
#Febonacci Series
term=int(input("Enter Range : ")) #10
if term<=0:
    print("Enter range greater than 0")
else:
    a=0
    b=1
    for i in range(term):
        print(a,end=" ")
        c=a+b
        a=b
        b=c
```
