```python
#Arithmatic Operator
#+,-,*,/,**,%,//
a=10
b=3
print(a+b)  #Addition
print(a-b)  #substraction
print(a*b)  #Multiplication
print(a/b)  #divison (gives float value)
print(a//b) #flooer divison(remove deciaml part)
print(a**b) #10**3 #exponant
print(a%b)  #modulas(gives reminder)
```

```python
#Assignment Operator
#it is used to assign value to variable
#=,+=,-=,*=,/=,%=,//=,**=

a=10
a+=5
print(a)

b=20
b-=10
print(b)

c=30
c*=3
print(c)
```

```python
#Comparison Operator
#It compares Value and give True or False

a=10
b=5
print(a==b)  #flase
print(a!=b)  #true
print(a>b)   #true
print(a<b    
print(a>=b)
print(a<=b)
```

```python
#Logical Operator
#and,or,not

a=10
b=20
c=30

print(a>b and b>c)  #false and false   #false
print(a<b and b>c)  #True and false    #false
print(a<b or b>c)   #true or false     #true
print(not(a>b))     #not(False)        #True
```

```python
#Membership Operator
#in,not in
#always return TRUE/False

print("a" in "Gagan")
print("a" not in "gagan")
```

```python
#Identity Operator
#is,is not
#It compares address of the variable
#comparison operator(==) compare values and identity operators compares address of value

a=[10,20]
b=[10,20]
print(a==b)   #True
print(a is b) #false
print(id(a))
print(id(b))
c=b
print(b is c) #true
print(b is not c)  #false
```

```python
#bitwise operator
#bin()- it is used to find binary of any number
#f-string- it is used to access variable in string

a=5
b=3
print(f"binary of {a} is ",bin(a))
print(f"binary of {b} is ",bin(b))
print(a&b) #binary and operation
print(a|b) #binary OR operation
print(a^b) #binary XOR operation
print(~5)
```

```python
#DATA TYPE-----

#numeric Data type
##INteger : -infinity to +infinity,discrete number,unindexed,immuatable
##Float: -infinite to + infinite,continous number

###type()- it is used to check datatype of variable
a=10
print(type(a))

b=10.5
print(type(b))
```

```python
#string DATA TYPE
##definition-- character inside double quote or single quote
##single line -- single quote (' ') or double quote(" ")
##multiple line--(''' ''') or (""" """)
#it is indexed,immutable,hetereogenious

s1="Gagan123"  #single line string
s2="""
    My
    Name
    Is
    Gagan
    """
print(s1)
print(s2)
```

```python
#sequence
#List- collection of element inside round bracket heterogenious data types with comma sperate values
## indexed,mutable,sequence
a=[10,20,10.5,"Gagan"]
print(a)
print(a[3])      #positive indexing
print(a[-1])     #negative indexing
a[2]=50
print(a[3][2])
```

```python
#TUPLE---it is collection of elements inside round bracket () with comma seperate.
##it is indexed,immutabel,sequence
### it execute faster than list
#we can remove elements from the tuple after creation but can be added.

t1=()  #empty tuple
t2=tuple() #empty tuple
print(type(t1))
print(type(t2))

t=(10,20,10.5,"Gagan")
print(t)
print(type(t))
```

```python
#Mapping
#Dictionary
#it is collection of data in key and value pair  key:value
#Unindexed,mutable,key should be unique and data should be duplicate
d=dict()  #empty dictionary
print(d)
print(type(d))

a={"name":["Yash","Ruchi","Disha"],"age":22}
print(a)
print(a.get("name"))
print(a.get("age"))
```

```python
#SEt
#set,frozenset
##it is a collection of unique data,hetrogenious,unindexed,mutable and perform all set operation
#union,intersection
#it is created by using {},unordered
s=set()    #empty set
print(s)
print(type(s))

s={10,20,20,20,30,"Gagan"}
print(s)
#frozenset
#frozenset is immutable"
#it gives distinct value
fs=frozenset(s)
print(type(fs))
```
