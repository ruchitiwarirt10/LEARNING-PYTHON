```python
#11. Replace Bad Words
#Replace every occurrence of ‘bad’ with ’***’
#Concept:
#- replace()
sentence = input("Enter a sentence: ")


new_sentence = sentence.replace("bad", "***")

print(new_sentence)
```

```python
#12. Remove Extra Spaces Accept a sentence containing multiple spaces. Convert it into a properly spaced sentence.
#Example: Input : Python is awesome
#Output: Python is awesome
#Concepts:- split(),- join()
sentence = input("Enter a sentence: ")

words = sentence.split()

new_sentence = " ".join(words)

print(new_sentence)
```

```python
#13. File Extension Finder Accept a filename and display its extension.
#Example: Input : report.pdf
#Output: pdf
#Concept:- split(‘.’)

filename = input("Enter filename: ")

extension = filename.split(".")[-1]

print("Extension:", extension)
```
