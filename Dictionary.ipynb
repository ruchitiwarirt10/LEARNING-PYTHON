```python
#DICTIONARY-data store in key:value pair
#it is mutable,ordered,inidexed
#key should be unique(integer,string,tuple)
#Value should be duplicate

d=dict()  #1st method
e={"name":"Gagan","age":32} #2nd method  e={key:value}
print(type(d))
print(type(e))
```

```python
#Access values of dictionary
e={"name":"Gagan","age":32}
print(e["name"])
print(e["age"])

#Adding new key value pair in dictionary
e["course"]="DA"    #new key:value pair added
print(e)

#Update value of key
e["age"]=30         #update value of key
print(e)
```

```python
#Value should be duplicate in dictionary
marks={"maths":90,"science":90,"english":90}
print(marks)

student={"name":"Bhavya","name":"Ruchi"}
print(student)         #if we create same key than last key overwrite

a={10:"A",20:"B",30:"C"}    #here key is integer
print(a)
```

```python
#Nested Dictionary-dictionary inside dictionary
#JSON format-java script object Notation
employees={
    101:{"name":"Bhavya",
         "salary":50000,
         "designation":"DA"},
    102:{"name":"Ruchi",
         "salary":50000,
         "designation":"DA"},
    103:{"name":"Disha",
         "salary":50000,
         "designation":"DA"},
}
print(employees)
```

```python
#Functions in dictionary-
#len(),max(),min(),sorted(),sum()
#all functions works on keys
#element=key:value
e={"name":"Gagan","age":32,"course":"DA"}
print(len(e))  #it gives total element in dictionary
print(max(e))  #if there are not integer than compare ASCII value
print(min(e))  #if there are not integer than compare ASCII value
print(sum(e))  #sum function only works when there are integer otherwise error occure
print(sorted(e))

```

```python
e={10:"Gagan",30:32,20:"DA"}
print("Length",len(e))  #it gives total element in dictionary
print("Maximum",max(e))  #if there are not integer than compare ASCII value
print("Minimum",min(e))  #if there are not integer than compare ASCII value
print("Sum",sum(e))  #sum function only works when there are integer otherwise error occure
print(sorted(e))     #it gives arranges list of key
print(sorted(e,reverse=True))  #reverse list of key

```

```python
#Methods in dictionary
#keys()-It gives keys of dictionary
#Values()-It gives values in dictionary
student={"name":"Gagan","age":32,"course":"DA"}
print(student.keys())
print(student.values())
print(student.items())
```

```python
#Looping in Dictionary-
student={"name":"Gagan","age":32,"course":"DA"}

for i in student:  #to get values of dictionary using loop
    print(i,end=" ")

print()

for i in student.values(): #it get keys of dictionary using loop
    print(i,end=" ")

print()

for i in student.items():   #to get both key and value of 
                            #dictionary using loop it return output as a tuple
    print(i)
```
