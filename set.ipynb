```python
#question1-create a set of five integers
s={10,20,20,30,40,40,50,50}
print(s)
print(type(s))
```

```python
#question2-create an empty set and print its type
s=set()
print(type(s))
```

```python
#question3-Add a new element to a set using add()
s={1,2,3,4,5}
s.add(6)
print(s)
```

```python
#question4-Add multiple elements using update()
s={1,2,3,4,5}
s.update({6,7})
print(s)
```

```python
#question5-Remove an element using remove()
s={1,2,3,4,5,6,7}
s.remove(6)
print(s)
```

```python
#question6-remove an element using discard()
s={1,2,3,4,5,6,7}
s.discard(5)
print(s)
```

```python
#question7-remove and print a random element using pop()
s={1,2,3,4,5,6}
s.pop()
print(s)
```

```python
#question8-find the length of a set
s={10,20,20,30,40,40,50,50}
print(len(s))
```
