```python
#wap to print first n natural number take n as a user input at run time?
n=int(input("Enter any number: "))
for i in range(1,n+1):
    print(i)
```

```python
#Wap to print first n even number and first n Odd number take n as an user input at run time?
n=int(input("Enter any number: "))
for i in range(1,(2*n)+1):
    if i%2==0:
        print(i,"even number")
    else:
        print(i,"odd number")
```

```python
#wap to print sum of first n natural number.?
n=int(input("Enter Any number: "))
sum=0
for i in range(1,n+1):
    sum=sum+i
print(sum)
```

```python
#wap to print sum of n even number and n odd number?
n=int(input("Enter any number: "))
sum_even=0
sum_odd=0
for i in range(1,(2*n)+1):
    if i%2==0:
        sum_even=sum_even+i
    else:
        sum_odd=sum_odd+i
print("sum of first",n,"even number",sum_even)
print("sum of first",n,"odd number",sum_odd)
    
```

```python
#wap to print factorial of n number.?
n=int(input("Enter any number: "))
fact=1
for i in range(1,n+1):
    fact=fact*i
print(fact)
```

```python
#wap to print fibonacci series up to nth term take n as user input at run time?
a=0
b=1
n=int(input("how many term: "))
print(a,b,end=" ")
for i in range(2,n):
    c=a+b
    print(c,end=" ")
    a=b
    b=c
```

```python
## wap to print nth term of fibonacci series
a=0
b=1
n=int(input("how many terms: "))
if n==1:
    print(a)
elif n==2:
    print(b)

elif n>2:
    for i in range(2,n):
        c=a+b
        a=b
        b=c
    print(c)
else:
    print("invalid input")
```

```python
#wap to check whether a given number is prime number of not.?
n=int(input("enter number: "))
if n==0 or n==1:
    print(n,"not a prime")
elif n>1:
    for i in range(2,n):
        if n%i==0:
            print(n,"not a prime")
            break
    else:
        print(n,"prime number")
else:
    print(n,"not a prime")
```

```python
#wap to print prime number between start and end take start and end as a user input at a run time.?
start=int(input("Enter any number: "))
end=int(input("Enter any number: "))
for n in range(start,end+1):
    if n>1:
        for i in range(2,n):
            if n%i==0:
                break
        else:
            print(n,end=" ")
```
