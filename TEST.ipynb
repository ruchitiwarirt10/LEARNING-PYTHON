```python
#Q1. Positive, Negative or Zero
#Write a Python program that takes an integer from the user and checks
#whether the number is:
#- Positive
#- Negative
#- Zero
#Example:
#Input : -8
#Output : Negative Number

num = int(input("Enter a number: "))
if num > 0:
    print("Positive Number")

elif num < 0:
    print("Negative Number")

else:
    print("Zero")
```

```python
#Q2. Sum of Digits
#Write a Python program to take an integer from the user and calculate
#the sum of all its digits.
#Example:
#Input : 12345
#Output : Sum of digits = 15

num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
#Q3. Reverse a String
#Write a Python program to take a string from the user and print the
#string in reverse order.Without using any method or slicing
#Example:
#Input : Python
#Output : nohtyP

text = input("Enter a string: ")

reverse = ""

for char in text:
    reverse = char + reverse

print("Reverse =", reverse)

```

```python
#Q4. Count Even and Odd Numbers in a List

#Write a Python program that takes a list of integers and counts how many
#numbers are even and how many numbers are odd.
#Example:
#Input : [10, 15, 22, 7, 8, 11]
#Output :
#Even numbers = 3
#Odd numbers = 3

numbers = [10, 15, 22, 7, 8, 11]

even = 0
odd = 0

for num in numbers:

    if num % 2 == 0:
        even = even + 1

    else:
        odd = odd + 1

print("Even numbers =", even)
print("Odd numbers =", odd)
```

```python
#Q8. Reverse Number

#Write a Python program to reverse number.
#Example:
#Input : 12345
#Output : 54321

num = int(input("Enter a number: "))

reverse = 0

while num > 0:

    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse =", reverse)
```
