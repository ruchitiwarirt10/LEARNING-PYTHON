## Operators

# Arithematic Operators-- + , - , * , / , // , % , **

```python
a=10
b=2
print("addition:-",a+b)
print("subtraction:-",a-b)
print("multiply:-",a*b)
print("simple division:-",a/b)
print("floor division:-",a//b)
print("modulas:-",a%b)
print("exponential power:-",a**b)
```

# Assignment operator :- =,+=,-=,*=.....

```python
a=10
b=5
c=a+b
a+=10     # a=a+10
b-=2      # b=b-2
c*=5      # c=c*5
a/=b      # a=a/b
c=a+b
print(a,b,c)
```

# Comparision Operator :- ==,!=,>=,<=,>,<

```python
a=10
b=5
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
print(a>b)
print(a<b)
```

# Logical Operator

```python
a=10
b=8
c=12
print(a>b and b>c)
print(a>b or b>c)
print(a>b and not b>c)
```

# Identity Operator :- is , is not

```python
# id() :- it is a predefined function that return memory address of given object.
a=10
b=8
print(id(a))
print(id(b))
print(a is b)
print(a is not b)
```

```python
c=20
d=20
print(id(c))
print(id(d))
print(c is d)
print(c is not d)
```

# Membership Operator :- in , not in

```python
a="hello"
print("h" in a)
print("h" not in a)
print("O" in a)
```

# Bitwise Operator :- &,|,^

```python
a=10
b=8
print(a&b)
print(a|b)
print(a^b)
```

```python
# bin() :- it is a predefined function always return binary of given integer object.

a=10
b=8
print(bin(a))
print(bin(b))
print("--------")
print(bin(a&b))
print(a&b)
print("-*-*-*-*-")
print("-*-*-*-*-")
print(bin(a))
print(bin(b))
print("--------")
print(bin(a|b))
print(a|b)
print("-*-*-*-*-")
print("-*-*-*-*-")
print(bin(a))
print(bin(b))
print("--------")
print(bin(a^b))
print(a^b)
```
