```python
#Wap function that return an armstrong number range between start and stop?(take 2 parameter start and stop)?
def arm(start,stop):
    l=[]
    for i in range(start,stop):
        a=len(str(i))
        sm=0
        old=i
        while i!=0:
            dig=i%10
            sm=sm+(dig**a)
            i=i//10
        if sm==old:
            l.append(old)
    return l
    
    
    
    
```
