```python
# Recursion :- It is a programming technique ,In which function call itself to solve smaller version of
# same problem until it reach its stop condition.(Base condition)

# def function_name(parameter):
#    if condition: } base condition
#       statement    } (stop condition)
#    return function_name() -----> recursive call

def print_number(n):
    if n<=1:
        return 1     # Base condition / Stop condition
    print(n)
    return print_number(n-1)

print(print_number(5))
```

```python
# forward
# None :- Garbage value

def print_number(n):
    if n==0:
        return 0
    print_number(n-1)
    print(n)

print(print_number(5))
```

```python
# Question :- Sum of n Natural numbers using recursion 

def sum_num(n):
    if n==0:
        return 0          # base condition
    return n+sum_num(n-1)
print(sum_num(10))
```

```python
# Factorial series using recursion (input=10)
# 10*9*8*.......*1

def factorial_num(n):
    if n==0:
        return 1
    return n*factorial_num(n-1)
print(factorial_num(10))
```

```python
# Febonacci series using recursion

def feb(n):
    if n<=1:
        return n
    return feb(n-1)+feb(n-2)

for i in range(10):
    print(feb(i),end=" ")
```

```python
# Check Palidrome String using recursion

def palindrome(s):
    if len(s)==1:
        return True
    if s[0]!=s[-1]:
        return False
    return palindrome(s[1:-1])

print(palindrome("madam"))
print(palindrome("gagan"))
```
