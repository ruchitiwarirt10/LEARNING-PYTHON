```python
#CONTROL FLOW FUNCTION DECIDES IN WHICH ORDER PYTHON EXECUTE ITS STATEMENT.
##2 TYPES OF CONTROL FLOW-
#1.CONDITIONAL STATEMENT,2.LOOPING STATEMENT

#1.CONDITIONAL STATEMENT-if,if else,if elif else,nested if

#IF BLOCK-
##if block only works for True conditions-

age=int(input("enter age : "))

if age>=18:
    print("eligibale for vote")
```

```python
#IF ELSE BLOCK-
##IF block works on TRUE and ELSE block works on FLASE condition

age=int(input("enter age : "))

if age>=18:
    print("eligibale for vote")

else:
    print("Not Eligible for vote")
```

```python
#IF ELIF ELSE-
##if there are multiple true conditions then we use this block-

marks=int(input("Enter Marks : "))
if marks>=85:
    print("A Grade")
elif marks>=70 and marks<85:
    print("B Grade")
elif marks>=50 and marks<70:
    print("C Grade")
elif marks>=35 and marks<50:
    print("D Grade")
elif marks<0:
    print("PLEASE ENTER VLID MARKS")
else:
    print("Fail")
```

```python
##NESTED IF-
##if block inside if block

age=int(input("ENTER AGE : "))
citizenship=input("Enter Citizenship : ")

if age>=18:
    if citizenship=="indian":
        print("YOU ARE eligible for Vote")
    else:
        print("indian citizenship required for vote")  #inner if else block
else:
    print("Age should be greater than 18")   #outer if else block


age=int(input("ENTER AGE : "))
citizenship=input("Enter Citizenship : ")

if age>=18 and citizenship=="indian":   #TRUE and TRUE=TRUE
        print("YOU ARE eligible for Vote")
else:
    print("YOU ARE NOT ELIGIBLE FOR VOTE")
```
