```python
"""Question 1: Student Class
 
  Concept: Class & Object

  Create a class named Student. 
  Requirements: - Instance variables: name,age 
                - Method: display() to print student details. 
                - Create one object and call the display() method."""

class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    def display(self):
        print("Student name ",self.name)
        print("Student age",self.age)
    
s1=Student("Ruchi",31)
s1.display()
```

```python
""" Question 2: Rectangle Class

  Concept: Class & Object

  Create a Rectangle class. 
  Requirements: - Variables: length, width 
                - Method: area() to calculate and return the area. 
                - Create an object and print the area."""

class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width

    def area(self):
        return self.length *  self.width

r1=Rectangle(20,30)
print("Area of Rectangle =",r1.area())
    
```

```python
"""Question 3: Employee Class

  Concept: Class & Object

  Create an Employee class. 
  Requirements: - Variables: emp_name, salary
                - Method: display() - Create two Employee objects and display their information."""

class Employee:
    def __init__(self,emp_name,salary):
        self.emp_name=emp_name
        self.salary=salary

    def display(self):
        print("employee name ",self.emp_name)
        print("employee salary ",self.salary)

e1=Employee("Ruchi",20000)
e1.display()

e2=Employee("Nidhi",30000)
e2.display()

```

```python
""" Question 4: Car Class

  Concept: Class & Object

  Create a Car class. 
  Requirements: - Variables: company, model, price 
                - Method: showDetails() - Create three car objects and display their details."""

class Car:
    def __init__(self,company,model,price):
        self.company=company
        self.model=model
        self.price=price

    def showDetails(self):
        print("car company ",self.company)
        print("car model ",self.model)
        print("cae price ",self.price)

c1=Car("Verna",11,1099000)
c1.showDetails()

c2=Car("Sierra",12,1149000)
c2.showDetails()

c3=Car("Defender",13,10700000)
c3.showDetails()
    
```

```python
""" Question 5: Bank Account

  Concept: Encapsulation

  Create a BankAccount class. 
  Requirements: - Private variable: __balance 
                - Methods: deposit(amount) ,withdraw(amount) ,displayBalance() 
                - Perform deposit and withdrawal operations."""

class BankAccount:
    def __init__(self,balance):
        self.__balance=balance

    def displaybalance(self):
        print("balance is ",self.__balance)

    def setbalance(self,amount):
        self.__balance=amount

a1=BankAccount(50000)
a1.displaybalance()
a1.setbalance(20000)
```

```python
"""Question 6: ATM Machine

  Concept: Encapsulation

  Create an ATM class. 
  Requirements: - Private variable: __pin 
                - Methods: changePin(newPin),showPin() 
                - Create an object and change the PIN."""

class ATM:
    def __init__(self,pin):
        self.__pin=pin
    def showpin(self):
        print("oldpin is ",self.__pin)
        
    def changepin(self,newpin):
        self.__pin=newpin

    def newpin(self):
        print("newpin is ",self.__pin)
p1=ATM(1234)
p1.showpin()
p1.changepin(4321)
p1.newpin()
```
