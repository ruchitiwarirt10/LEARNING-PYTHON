```python
# Lamda Function :-
# It is an anonymous function
# It is a function without name
# Syntax lamda argument:expression
# It is used for small task.
# we write code in single line using lambda function 

def square(n):
    return n*n

print(square(5))

# square of number using lambda function
# example of lambda function with single argument

square= lambda x:x*x       # lambda arguments:expression
print(square(4))
```

```python
# lambda function with multiple arguments
# Question : Sum of two numbers

result= lambda a,b:a+b
print(result(10,20))
```

```python
# subtraction of two numbers

result= lambda a,b:a-b
print(result(10,5))

# multiplication of two numbers

result1= lambda a,b:a*b
print(result1(5,8))
```

```python
# sum of three numbers

result= lambda a,b,c:a+b+c
print(result(5,6,10))
```

```python
# convert string to uppercase

upper= lambda s:s.upper()
print(upper("gagan"))
```
