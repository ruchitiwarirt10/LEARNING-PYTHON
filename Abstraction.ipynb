```python
#Abstraction-in abstraction we hide implementation details and show only relevenent data to user.
            #Abstraction also used to enforce some rules to child class.

#for abstraction we use "abc" module   #Example-abc.py  -python file is also called module.

#1.  abc= this is module to achive abstraction.
#2. ABC = Abstract Base Class =It provides common structure to child classes
#3. abstractmethod = This is decorator and it is used to create method that should be redefine in child class

from abc import ABC,abstractmethod

class Bank(ABC):         #Abstract Class
    @abstractmethod
    def withdraw(self):  #Abstract method
        pass

class SbiBank(Bank):
    def __init__(self,balance):
        self.__balance=balance      #private variable

    def withdraw(self,amount):
        self.__balance-=amount
        print("Current Balance after withdrawl of ",amount,"is",self.__balance)

    def displaybalance(self):
        print("Balance is ",self.__balance)

class UbiBank(Bank):
    pass                           #Error here because withdrawal method is not defined

        
    

s=SbiBank(50000)
s.displaybalance()
s.withdraw(30000)
u=UbiBank()

    
```

```python
#3 Types of Methods-
#1.Instance/Normal methods
#2.Class method-We Use @classmethod decorator to create class method and pass 'cls' as argument in that method.
#3.Static Method-We use @staticmethod decorator to access normal function outside class

class Bank:
    bank_name="This is State bank of India"     #Class Variable
    def __init__(self,balance):
        self.__balance=balance      #private # Instance variable  

    def withdraw(self,amount):
        self.__balance-=amount
        print("Current Balance after withdrawl of ",amount,"is",self.__balance)

    def displaybalance(self):
        print("Balance is ",self.__balance)

    @classmethod
    def displayBank(cls):
        print("Bank name is ",cls.bank_name)

    def add(a,b):
        c=a+b
        print(c)                #Static Variable

s=Bank(50000)
s.displayBank()
```

```python
class Bank:
    bank_name="This is State bank of India"     #Class Variable
    def __init__(self,balance):
        self.__balance=balance      #private # Instance variable  

    def withdraw(self,amount):
        self.__balance-=amount
        print("Current Balance after withdrawl of ",amount,"is",self.__balance)

    def displaybalance(self):
        print("Balance is ",self.__balance)

    @classmethod
    def displayBank(cls):
        print("Bank name is ",cls.bank_name)
    
    @staticmethod               #Static method
    def add(a,b):
        c=a+b
        print(c)                #Static Variable

s=Bank(50000)
s.displayBank()
s.add(10,20)
```
