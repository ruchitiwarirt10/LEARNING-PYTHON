```python
"""Q3. arange() + reshape() — Logic

Create an array containing numbers from 10 to 69 using arange().
Reshape it into a 3D array of shape (4, 5, 3).
Then:
1. Find the sum of the elements along the last axis.--(X)--
2. Find the maximum value from each 5 × 3 array.--(X)--
3. Extract the third 5 × 3 array.--(X)--
4. Replace all values divisible by 7 with -1.--(X)--"""

arr=np.arange(10,70).reshape(4,5,3)
print(arr)
arr_new=np.sum(arr,axis=2)
print(arr_new)
arr_new=np.max(arr,axis=0)
print(arr_new)
```

```python
#Q5. Random Array + Quartiles
#Generate a random NumPy array containing 50 integers between 10 and 100.
#Find:
#• Q1
#• Median
#• Q3
#• IQR
#• Lower Fence
#• Upper Fence
arr=np.random.randint(10,100,50)
print(arr)
arr1=np.median(arr)
print(arr1)
```

```python
Q1=np.quantile(arr,0.25)
Q1
```

```python
Q2=np.quantile(arr,0.50)
Q2
```

```python
Q3=np.quantile(arr,0.75)
Q3
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
Upper_fence=Q3+(1.5*IQR)
Upper_fence
```

```python
Lower_fence=Q1-(1.5*IQR)
Lower_fence
```
